```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
import pandas as pd

df = pd.read_csv("Retail Sales Data Set.csv")

df.head()
```

```python
# Dataset information

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Descriptive Statistics

print("Mean:")
print(df.mean(numeric_only=True))

print("\nMedian:")
print(df.median(numeric_only=True))

print("\nMode:")
print(df.mode(numeric_only=True).iloc[0])

print("\nStandard Deviation:")
print(df.std(numeric_only=True))
```

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Create Month column
df['Month'] = df['Date'].dt.to_period('M').astype(str)

# Calculate monthly sales
monthly_sales = df.groupby('Month')['Total Amount'].sum()

# Display monthly sales
print(monthly_sales)

# Plot monthly sales trend
plt.figure(figsize=(12, 6))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Gender-wise sales analysis

gender_sales = df.groupby('Gender')['Total Amount'].sum()

print("Sales by Gender:")
print(gender_sales)

# Bar chart
plt.figure(figsize=(7, 5))
gender_sales.plot(kind='bar')

plt.title('Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
print(df.columns.tolist())
```

```python
# Revenue by Product Category

category_revenue = df.groupby('Product Category')['Total Amount'].sum().sort_values(ascending=False)

print("Revenue by Category:")
print(category_revenue)

# Plot
plt.figure(figsize=(8, 5))
category_revenue.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Sales distribution by Product Category

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Product Category',
    y='Total Amount'
)

plt.title('Sales Distribution by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Amount')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Key findings for business insights

print("Total Revenue:", df['Total Amount'].sum())
print("Total Quantity Sold:", df['Quantity'].sum())

print("\nBest Selling Category:")
print(df.groupby('Product Category')['Quantity'].sum().sort_values(ascending=False))

print("\nHighest Revenue Category:")
print(df.groupby('Product Category')['Total Amount'].sum().sort_values(ascending=False))

print("\nAverage Order Value:")
print(df['Total Amount'].mean())

print("\nGender-wise Revenue:")
print(df.groupby('Gender')['Total Amount'].sum())

print("\nMonthly Sales:")
print(monthly_sales)
```

```python
# Key Observations

1. The total revenue generated from the retail sales dataset is ₹456,000, with a total quantity sold of 2,514 units.

2. Clothing is the best-selling product category by quantity, with 894 units sold.

3. Electronics generates the highest revenue at ₹156,905, indicating strong revenue contribution from this category.

4. The average order value is ₹456, showing the average revenue generated per transaction.

5. Female customers generated ₹232,840 in revenue, while male customers generated ₹223,160. Female customers contributed slightly higher revenue.

6. Monthly sales show variations across the year, with October 2023 recording the highest visible monthly sales of ₹46,580.

# Business Recommendations

1. Focus on Electronics:
   Since Electronics generates the highest revenue, the business should maintain sufficient inventory and promote high-value electronic products.

2. Promote Clothing:
   Clothing has the highest sales volume. The business can use discounts, bundles, and cross-selling offers to increase revenue from this category.

3. Target Customer Segments:
   Female customers generate slightly higher revenue. Marketing campaigns can be personalized based on customer purchasing behaviour while continuing to target all customer groups.

# Conclusion

The EDA provides useful insights into sales performance, customer behaviour, product categories, and monthly sales trends. Electronics is the strongest revenue-generating category, while Clothing has the highest sales volume. These findings can help the business improve inventory planning, marketing strategies, and sales performance.
```
